```python
import os
from analysis import (
    compute_simulation_statistics,
    run_batch_and_collect_traces,
)
```

```python
print(
    "🚀 Running Fast Minimax (P1) vs Greedy (P2) Verification (1000 Games)..."
)
raw_results_m_v_g, _ = run_batch_and_collect_traces(
    num_games=1000, 
    target_score=3,
    p1_type="minimax",
    p2_type="greedy",
    force_overwrite=True,
)

stats_m_v_g = compute_simulation_statistics(raw_results_m_v_g)
print(f"P1 (Minimax) Win Rate: {stats_m_v_g['p1_win_rate'] * 100:.1f}%")
print(f"P2 (Greedy)  Win Rate: {stats_m_v_g['p2_win_rate'] * 100:.1f}%")
print(f"Draw Rate:           {stats_m_v_g['draw_rate'] * 100:.1f}%")
```

```python
import seaborn as sns
import matplotlib.pyplot as plt

# 2. Compact single-line visualization of outcome comparison
fig, ax = plt.subplots(figsize=(8, 4))
sns.barplot(
    x=["P1 Win", "P2 Win", "Draw"],
    y=[
        stats_m_v_g["p1_win_rate"] * 100,
        stats_m_v_g["p2_win_rate"] * 100,
        stats_m_v_g["draw_rate"] * 100,
    ],
    palette=["#2ecc71", "#e74c3c", "#95a5a6"],
    ax=ax,
)
ax.set_title(
    "Outcome Share: Minimax (P1) vs Greedy (P2) [Target Score = 3]",
    fontweight="bold",
)
ax.set_ylabel("Percentage (%)")
for p in ax.patches:
  ax.annotate(
      f"{p.get_height():.1f}%",
      (p.get_x() + p.get_width() / 2.0, p.get_height()),
      ha="center",
      va="bottom",
  )
plt.tight_layout()
plt.show()
```
